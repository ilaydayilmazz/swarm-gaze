# Bee Gaze

**What does a machine notice when it looks at a work of art?**

1. The machine looks at ~100 public-domain works by one artist and learns that artist's visual vocabulary (a set of typical marks).
2. It looks at a new image and scores each region by how much it resembles the artist's signature marks.
3. A colony of artificial bees (Artificial Bee Colony, ABC) forages over that map. Wherever a bee feeds, it paints one of the learned marks.
4. The result is a new image made only from what the machine learned.
